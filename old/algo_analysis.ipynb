## CB 題庫


```python
# 紀錄次數 = 迭代次數次(初始解)
run_time = 20 # 執行次數
directory = ["OR5x100","OR5x250","OR5x500","OR10x100","OR10x250","OR10x500"]
problem = ["1","2","3","4","5"]
algorithm = "BSCA_V1_25"

result_table = np.zeros([len(problem)*len(directory),run_time],dtype=int)

# 循環瀏覽您的 CSV 文件
for j, file in enumerate(directory):
    result = np.zeros([len(problem),run_time])
    for i in range(run_time):
        for k, prob in enumerate(problem):
            
            best_df_name = f"runtime_{i}_{file}-0.25_{prob}_algo_{algorithm}.csv" # 最佳解
            df_path = f"output4/{algorithm}/cb/{file}/{best_df_name}" # 最佳解

            with open(df_path, 'r') as csvfile:

                text = csvfile.read() # 讀取
                
                if text != None:
                    match = re.search(r'(\d+\.\d+)', text).group(1)
                    result[k,i] = float(match)
                else:
                    result[k,i] = ""
    # print(result)
    result_table[j*len(problem):j*len(problem)+5] = result
        

df = pd.DataFrame(result_table)
df
```

```python
# 計算 PDev
def calculate_PDev(data):
    opt = [
    24381, 24274, 23551, 23534, 23991, 59312, 61472, 62130, 59463, 58951,
    120148, 117879, 121131, 120804, 122319, 23064, 22801, 22131, 22772, 22751,
    59187, 58781, 58097, 61000, 58092, 117821, 119249, 119215, 118829, 116530] # 理論最佳
    PDev = np.zeros([len(opt)])
    # print(data)
    for i,d in enumerate(data):
        PDev[i] = (opt[i] - d ) / opt[i] * 100 # 計算值
    print(data[0])
    return PDev

row_max = df.max(axis=1) # 計算平均值
row_means = df.mean(axis=1) # 計算平均值
row_min = df.min(axis=1) # 計算平均值
row_std = df.std(axis=1)
row_PDev = calculate_PDev(row_means) # PDev 值            

# print(row_max)
# 將結果存儲在新的 DataFrame 中
# 將結果合併到一個新的 DataFrame 中
static_df = pd.DataFrame({'Best': row_max, 'Mean': row_means, 'Worst':row_min,"Std":row_std,"PDev":row_PDev})
# 創建多層次索引
static_df.columns = pd.MultiIndex.from_product([[f'{algorithm}'], static_df.columns])
# 查看結果
# 沿水平軸合併兩個 DataFrame
result_df = pd.concat([result_df, static_df], axis=1)
result_df

# df = pd.DataFrame(result_table)
# df
```

## GK 題庫

```python
# 計算 PDev
def calculate_PDev(data):
    opt = [3766,3958,5656,5767,7560,7677,19221,18806,58089] # 理論最佳
    PDev = np.zeros([len(opt)])
    # print(data)
    for i,d in enumerate(data):
        PDev[i] = (opt[i] - d ) / opt[i] * 100 # 計算值
    print(data[0])
    return PDev

# 紀錄次數 = 迭代次數次(初始解)
run_time = 20 # 執行次數
problem = ["01","02","03","04","05","06","07","08","09"]
algorithm = "BSMA_V1_008"

result_table = np.zeros([len(problem),run_time],dtype=int)
result_df = pd.DataFrame()
# 循環瀏覽您的 CSV 文件
for i in range(run_time):
    for j, prob in enumerate(problem):
        best_df_name = f"runtime_{i}_prob_{prob}_algo_{algorithm}.csv" # 最佳解
        df_path = f"output3/{algorithm}/GK/{best_df_name}" # 最佳解
        with open(df_path, 'r') as csvfile:
            text = csvfile.read() # 讀取
            if text != None:
                match = re.search(r'(\d+\.\d+)', text).group(1)
                result_table[j,i] = float(match)
            else:
                result_table[j,i] = ""

# 顯示數據細節
df = pd.DataFrame(result_table)
df
```

# 威爾斯

```python
import numpy as np
from scipy.stats import wilcoxon

# 假設這些數據是某種治療前後的血壓讀數
data_before = np.array([120, 125, 130, 140, 145, 150, 135, 160, 140, 155])
data_after = np.array([115, 123, 130, 138, 140, 148, 130, 155, 135, 150])

# 計算差異
data_diff = data_after - data_before

# 執行 Wilcoxon Signed-Rank Test
w_statistic, p_value = wilcoxon(data_diff)

print(f"Wilcoxon test statistic: {w_statistic}")
print(f"P-value: {p_value}")

# 判斷結果
if p_value < 0.05:
    print("There is a significant difference between the two conditions.")
else:
    print("There is no significant difference between the two conditions.")
```
