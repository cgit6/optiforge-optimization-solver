```python
import matplotlib.pyplot as plt
import numpy as np

# 盒鬚圖的名稱
names = ["BSMA", "BSCA", "BSMASCA", "HLMS"]

# 假設數據
data_info = [
    {"max": 14487.33, "min": 14472.33, "mean": 14479.73},
    {"max": 14485, "min": 14471.67, "mean": 14478.31},
    {"max": 14490.89, "min": 14477.22, "mean": 14484.91},
    {"max": 14426.11, "min": 14219.66, "mean": 14362.22}
]

# 處理數據以適應盒鬚圖的要求
processed_data = []
for info in data_info:
    # 每組數據只包含最大值、平均值和最小值
    processed_data.append([info["min"], info["mean"], info["max"]])

# 繪製盒鬚圖
fig = plt.figure(figsize =(10, 7))

# 這裡我們使用 'whis' 參數來設定盒頂和盒底的位置
# 因為數據中只有三個點（最小值、平均值、最大值），我們不使用 IQR 來計算盒頂和盒底
bp = plt.boxplot(processed_data, vert = 1, patch_artist = True, whis=[0, 100], showfliers=False)

# 添加每個盒鬚圖的名稱
for i, name in enumerate(names):
    plt.text(i + 1, processed_data[i][2] + 1, name, ha='center', va='bottom', fontsize=10)

# 設置圖標題
plt.title('GK test')
plt.show()
```

```python
import matplotlib.pyplot as plt

# 繪製帶有名稱的長條圖

# 假設的三個值及其名稱
values = [4.02, 3.68, 3.22]
names = ["BSMA", "BSCA", "BSMASCA"]

# 調整長條的寬度

# 設定長條的寬度
bar_width = 0.5

# 繪製長條圖
fig, ax = plt.subplots()
bars = ax.bar(names, values, width=bar_width)

# 在每個長條上添加名稱
for bar in bars:
    height = bar.get_height()
    ax.annotate(f'{height}',
                xy=(bar.get_x() + bar.get_width() / 2, height),
                xytext=(0, 3),  # 3點的偏移量
                textcoords="offset points",
                ha='center', va='bottom')

plt.show()
```
